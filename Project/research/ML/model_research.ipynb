```python
data['Datetime'] = pd.to_datetime(data['Datetime'])

# Определяем дату начала тестовой выборки (последний месяц)
test_start_date = data['Datetime'].max() - pd.DateOffset(months=1)

# Определяем дату начала валидационной выборки (предпоследний месяц)
val_start_date = data['Datetime'].max() - pd.DateOffset(months=2)

# Разделение данных на тренировочную, валидационную и тестовую выборки по времени
train_data = data[data['Datetime'] < val_start_date]  # все, что до предпоследнего месяца
val_data = data[(data['Datetime'] >= val_start_date) & (data['Datetime'] < test_start_date)]  # предпоследний месяц
test_data = data[data['Datetime'] >= test_start_date]  # последний месяц

# Признаки, которые будем использовать
features = new_columns

# Разделение на признаки (X) и целевую переменную (y) для каждой выборки
X_train = train_data[new_columns]
y_train = train_data['target']

X_val = val_data[new_columns]
y_val = val_data['target']

X_test = test_data[new_columns]
y_test = test_data['target']
```

```python
from sklearn.metrics import precision_score, recall_score, accuracy_score, f1_score


def calculate_metrics_table(y_true, y_pred_prob, thresholds=[0.5, 0.6, 0.7, 0.8]):
    metrics_table = []

    for threshold in thresholds:
        y_pred = (y_pred_prob >= threshold).astype(int)
        metrics = {
            'Cutoff': threshold,
            'Precision': precision_score(y_true, y_pred),
            'Recall': recall_score(y_true, y_pred),
            'Accuracy': accuracy_score(y_true, y_pred),
            'F1-Score': f1_score(y_true, y_pred)
        }
        metrics_table.append(metrics)

    return pd.DataFrame(metrics_table)*100
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
import catboost as cb
from sklearn.metrics import roc_auc_score
from concurrent.futures import ThreadPoolExecutor

# 1. Применение StandardScaler к тренировочной, валидационной и тестовой выборкам
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
 
# Трансформируем валидационную и тестовую выборки
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

# Обучение модели model на обучающей выборки, predict_proba для предсказаний на валидационной и тестовой выборках
def train_and_predict(model, X_train, y_train, X_val, X_test):
    model.fit(X_train, y_train)
    pred_val = model.predict_proba(X_val)[:, 1]
    pred_test = model.predict_proba(X_test)[:, 1]
    return pred_val, pred_test, model.__class__.__name__

# Случайный лес с 100 деревьями и максимальной глубиной 5.
rf_model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
# Одно дерево решений с максимальной глубиной 4.
dt_model = DecisionTreeClassifier(max_depth=4, random_state=42)
# Логистическая регрессия с L2-регуляризацией и параметром C=10.0
lr_model = LogisticRegression(solver='saga', penalty='l2', C=10.0, random_state=42)

# Градиентный бустинг на деревьях 
cb_model = cb.CatBoostClassifier(
    n_estimators=1000,
    random_state=42,
    learning_rate=0.1,
    l2_leaf_reg=3.0,
    bagging_temperature=1.0,
    rsm=0.8,
    subsample=0.8,
    verbose=0
)

# Используем ThreadPoolExecutor для параллельного обучения моделей в разных потоках
with ThreadPoolExecutor() as executor:
    futures = [
        executor.submit(train_and_predict, rf_model, X_train_scaled, y_train, X_val_scaled, X_test_scaled),
        executor.submit(train_and_predict, dt_model, X_train_scaled, y_train, X_val_scaled, X_test_scaled),
        executor.submit(train_and_predict, lr_model, X_train_scaled, y_train, X_val_scaled, X_test_scaled),
        executor.submit(cb_model.fit, X_train_scaled, y_train, eval_set=(X_val, y_val), early_stopping_rounds=50)
    ]

    # Ждем завершения всех потоков
    results = [future.result() for future in futures[:-1]]  
    cb_future = futures[-1]

    cb_future.result()  
    cb_pred_val = cb_model.predict_proba(X_val_scaled)[:, 1]
    cb_pred_test = cb_model.predict_proba(X_test_scaled)[:, 1]
    results.append((cb_pred_val, cb_pred_test, 'CatBoostClassifier'))
    
for pred_val, pred_test, model_name in results:
    roc_auc = roc_auc_score(y_val, pred_val)
    # Для валидационной выборки
    print(f"{model_name} ROC AUC VAL : {roc_auc}")
    roc_auc = roc_auc_score(y_test, pred_test)
    # Для тестовой
    print(f"{model_name} ROC AUC TEST: {roc_auc}")
```

Далее с помощью optuna находим лучшие гиперпараметры 

```python
import optuna

# Функция для подбора гиперпараметров RandomForestClassifier
def objective_rf(trial):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 50, 500),
        'max_depth': trial.suggest_int('max_depth', 3, 20),
        'min_samples_split': trial.suggest_int('min_samples_split', 2, 10),
        'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 10),
        'max_features': trial.suggest_categorical('max_features', ['sqrt', 'log2', None]),
        'random_state': 42,
    }
    model = RandomForestClassifier(**params)
    model.fit(X_train, y_train)
    y_val_pred = model.predict_proba(X_val)[:, 1]
    auc = roc_auc_score(y_val, y_val_pred)
    return auc
```

```python
# Подбор гиперпараметров для RandomForestClassifier
study_rf = optuna.create_study(direction='maximize')
study_rf.optimize(objective_rf, n_trials=40, n_jobs=-1)
print("Лучшие параметры для RandomForestClassifier:", study_rf.best_params)
print("Лучший ROC AUC Score для RandomForestClassifier:", study_rf.best_value)
```

```python

# Функция для подбора гиперпараметров DecisionTreeClassifier
def objective_dt(trial):
    params = {
        'max_depth': trial.suggest_int('max_depth', 3, 20),
        'min_samples_split': trial.suggest_int('min_samples_split', 2, 10),
        'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 10),
        'max_features': trial.suggest_categorical('max_features', ['sqrt', 'log2', None]),
        'random_state': 42,
    }
    model = DecisionTreeClassifier(**params)
    model.fit(X_train, y_train)
    y_val_pred = model.predict_proba(X_val)[:, 1]
    auc = roc_auc_score(y_val, y_val_pred)
    return auc
```

```python
# Подбор гиперпараметров для DecisionTreeClassifier
study_dt = optuna.create_study(direction='maximize')
study_dt.optimize(objective_dt, n_trials=40, n_jobs=-1)
print("Лучшие параметры для DecisionTreeClassifier:", study_dt.best_params)
print("Лучший ROC AUC Score для DecisionTreeClassifier:", study_dt.best_value)
```

```python

# Функция для подбора гиперпараметров LogisticRegression
def objective_lr(trial):
    params = {
        'C': trial.suggest_loguniform('C', 1e-3, 10),
        'penalty': trial.suggest_categorical('penalty', ['l1', 'l2']),
        'solver': trial.suggest_categorical('solver', ['liblinear', 'saga']),
        'max_iter': trial.suggest_int('max_iter', 100, 1000),
        'random_state': 42,
    }
    model = LogisticRegression(**params)
    model.fit(X_train, y_train)
    y_val_pred = model.predict_proba(X_val)[:, 1]
    auc = roc_auc_score(y_val, y_val_pred)
    return auc
```

```python

# Подбор гиперпараметров для LogisticRegression
study_lr = optuna.create_study(direction='maximize')
study_lr.optimize(objective_lr, n_trials=40, n_jobs=-1)
print("Лучшие параметры для LogisticRegression:", study_lr.best_params)
print("Лучший ROC AUC Score для LogisticRegression:", study_lr.best_value)
```

```python

from catboost import CatBoostClassifier

def objective_cb(trial):
    params = {
        'iterations': trial.suggest_int('iterations', 100, 1000),
        'depth': trial.suggest_int('depth', 3, 10),
        'learning_rate': trial.suggest_loguniform('learning_rate', 0.01, 0.3),
        'l2_leaf_reg': trial.suggest_loguniform('l2_leaf_reg', 1e-3, 10),
        'random_seed': 42,
        'verbose': 0,
        'early_stopping_rounds': 50
    }
    model = CatBoostClassifier(**params)
    model.fit(X_train, y_train, eval_set=(X_val, y_val), early_stopping_rounds=50, verbose=0)
    y_val_pred = model.predict_proba(X_val)[:, 1]
    auc = roc_auc_score(y_val, y_val_pred)
    return auc

study = optuna.create_study(direction='maximize')
study.optimize(objective_cb, n_trials=50, n_jobs=-1)

print("Лучшие параметры:", study.best_params)
print("Лучший ROC AUC Score:", study.best_value)
```
