```python
from plotly.subplots import make_subplots
def create_candlestick_chart(df, title):

	fig = make_subplots(
		rows=2, 
		cols=1, 
		shared_xaxes=True, 
		vertical_spacing=0.02, 
		row_heights=[0.7, 0.3],
		subplot_titles=("Цена", "Объем")
	)

	fig.add_trace(
		go.Candlestick(
			x=df['Datetime'],
			open=df['Open'],
			high=df['High'],
			low=df['Low'],
			close=df['Close'],
			name="Цена"
		),
		row=1, col=1
	)

	fig.add_trace(
		go.Bar(
			x=df['Datetime'],
			y=df['Volume'],
			name="Объем",
			marker_color='rgba(0, 150, 255, 0.8)'
		),
		row=2, col=1
	)

	fig.update_layout(
		title=title,
		xaxis_rangeslider_visible=False,
	#	template="plotly_white",
		height=800,
		legend=dict(
			yanchor="top",
			y=0.99,
			xanchor="left",
			x=0.01
		)
	)

	if 'anomalies_price' in df.columns:
		price_anomalies = df[df['anomalies_price'] == 1]
		if not price_anomalies.empty:
			fig.add_trace(
				go.Scatter(
					x=price_anomalies['Datetime'],
					y=price_anomalies['Close'],
					mode='markers',
					marker=dict(
						symbol='circle',
						size=10,
						color='rgba(255, 0, 0, 0.8)',
						line=dict(width=2, color='rgba(255, 0, 0, 1)')
					),
					name='Аномалии цены'
				),
				row=1, col=1
			)

	if 'anomalies_volume' in df.columns:
		volume_anomalies = df[df['anomalies_volume'] == 1]
		if not volume_anomalies.empty:
			fig.add_trace(
				go.Scatter(
					x=volume_anomalies['Datetime'],
					y=volume_anomalies['Volume'],
					mode='markers',
					marker=dict(
						symbol='triangle-up',
						size=10,
						color='rgba(255, 165, 0, 0.8)',
						line=dict(width=2, color='rgba(255, 165, 0, 1)')
					),
					name='Аномалии объема'
				),
				row=2, col=1
			)
			
	# Настройка стилей осей
	fig.update_yaxes(title_text="Цена", row=1, col=1, gridcolor='rgba(230, 230, 230, 1)')
	fig.update_yaxes(title_text="Объем", row=2, col=1, gridcolor='rgba(230, 230, 230, 1)')
	fig.update_xaxes(title_text="Дата", row=2, col=1, gridcolor='rgba(230, 230, 230, 1)')

	fig.update_layout(
		plot_bgcolor='rgba(250, 250, 250, 1)',
		paper_bgcolor='rgba(250, 250, 250, 1)'
	)

	return fig
```
